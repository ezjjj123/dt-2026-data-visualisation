Create a chart showing who puts in the most effort. Effort can be find from number of hours. (For example Student A has the highest study hours meaning they put in the most effort.)

```python
import pandas as pd 
df= pd.read_csv("Competencies.csv")
df
```

```python
import matplotlib.pyplot as plt
avg_hours = df.groupby("Student")["Hours_Studied"].mean()
#colors= plt.cm.plasma(range(len(avg_hours)))
avg_hours.plot(kind="bar", x="Student", y="Hours_Studies", color="#FFC0CB")
plt.show()
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*50, alpha=0.6, color="#800080")
plt.xlabel("Hours studied")
plt.ylabel("Grades")
plt.title("Grade vs hours studies (Bubble size=hours)")
```

```python
plt.figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Disribution of study hours and grades")
```
