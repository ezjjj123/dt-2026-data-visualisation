Create chart showing who puts in the most effort. Effort can be find from number of hours. For example Student A has the highest study hours meaning they put in the most effort. 

```python
import pandas as pd 
df = pd.read_csv("Competencies.csv")
df
```

```python
import matplotlib.pyplot as plt
avg_hours = df.groupby("Student")["Hours_Studied"].mean()
colors= plt.cm.plasma(range(len(avg_hours)))
avg_hours.plot(kind="bar", x="Students", y="Hours_Studied", color="#FFC0CB")
plt.show()
```

```python
plt.scatter(df["Grade"], df["Hours_Studied"], s=df["Hours_Studied"]*25, alpha=0.4,color="#FFC0CB")
plt.title("Grade vs Hours Studied")
plt.xlabel("Grade")
plt.ylabel("Hours Studied")
```

```python
plt.Figure(figsize=(8,5))
df[["Grade", "Hours_Studied"]].boxplot()
plt.title("Box plot for grade and hours studied")
plt.show
```
